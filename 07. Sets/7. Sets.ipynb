### **Sets**

Sets are used to store multiple items in a single variable. Set items are unordered, unchangeable, and do not allow duplicate values.

- **Unordered**\
Unordered means that the items in a set do not have a defined order.

- **Unchangeable**\
Set items are unchangeable, meaning that we cannot change the items after the set has been created.

- **Duplicates Not Allowed**\
Sets cannot have two items with the same value.

**Creating a Set**

```python
set1 = {1, 2, 3, 4}
print(set1) 
```

**Length of a Set**

```python
# Get the number of items in a set

set1 = {"apple", "banana", "cherry", "mango"}
print(len(set1))
```

**Type Casting**

`set()` method is used to convert other data types, such as lists or tuples, into sets.

```python
set1 = set(["apple", "banana", "cherry", "mango"])
print(set1)
```

**Accessing Set Items**

```python
# Loop through the set, and print the values

set1 = {"apple", "banana", "cherry", "mango"}

for x in set1:
    print(x)
```

```python
# Check if "banana" is present in the set

set1 = {"apple", "banana", "cherry", "mango"}

if "banana" in set1:
    print("banana is present in the set")
    
print("-------------------------")

set1 = {"apple", "banana", "cherry", "mango"}
print("banana in set1")
```

**Add Set Items**

Once a set is created, you cannot change its items, but you can add new items.

```python
# Add an item to a set

set1 = {"apple", "banana", "cherry", "mango"}
set1.add("orange")
print(set1)
```

```python
# Add items from another set into the current set

set1 = {"apple", "banana", "cherry"}
set2 = {"pineapple", "mango", "papaya"}

set1.update(set2)
print(set1)
```

**Remove Set Items**

```python
# Remove "banana" from the set

# Using remove()

set1 = {"apple", "banana", "cherry"}
set1.remove("banana")
print(set1)

print("-------------------------")

# Using discard()

set1 = {"apple", "banana", "cherry"}
set1.discard("banana")
print(set1)
```

**Loop Sets**

```python
# Loop through the set, and print the values

set1 = {"apple", "banana", "cherry"}

for x in set1:
    print(x)
```

**Join Sets**

There are several ways to join two or more sets in Python.

- The `union()` and `update()` methods joins all items from both sets.

- The `intersection()` method keeps ONLY the duplicates.

- The `difference()` method keeps the items from the first set that are not in the other set(s).

- The `symmetric_difference()` method keeps all items EXCEPT the duplicates.

**1. Union** -> The union() method returns a new set with all items from both sets.

```python
# Join set1 and set2 into a new set

# Using union() method

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)

print("-------------------------")

# Using | operator

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1 | set2
print(set3)
```

- **Join Multiple Sets**

```python
# Using union() method

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

new_set = set1.union(set2, set3, set4)
print(new_set)

print("-------------------------")

# Using | operator

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

new_set = set1 | set2 | set3 | set4
print(new_set)
```

- **Join a Set and a Tuple**

```python
set1 = {"a", "b", "c"}
tuple1 = (1, 2, 3)

new_set = set1.union(tuple1)
print(new_set)
```

**2. Update** -> The update() method inserts all items from one set into another.

```python
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}

set1.update(set2)
print(set1)
```

**3. Intersection** -> The intersection() method will return a new set, that only contains the items that are present in both sets.

```python
# Using intersection() method

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3)

print("-------------------------")

# Using & operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 & set2
print(set3)
```

**4. Difference** -> The difference() method will return a new set that will contain only the items from the first set that are not present in the other set.

```python
# Using difference() method

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.difference(set2)
print(set3)

print("-------------------------")

# Using - operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 - set2
print(set3)
```

**5. Symmetrical Difference** -> The symmetric_difference() method will keep only the elements that are NOT present in both sets.

```python
# Using symmetric_difference() method

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.symmetric_difference(set2)
print(set3)

print("-------------------------")

# Using - operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 - set2
print(set3)
```

### **Frozenset**

Frozenset is an immutable version of a set. Its elements cannot be changed after creation, but you can perform operations like union, intersection and difference.

**Creating a frozenset**

```python
x = frozenset({"apple", "banana", "cherry"})
print(x)
```

### **Set Methods**

**1. add()** -> Adds an element to the set

```python
# Add an element to the set

set1 = {"apple", "banana", "cherry"}
set1.add("orange")
print(set1)

print("-------------------------")

# Try to add an element that already exists

set1 = {"apple", "banana", "cherry"}
set1.add("apple")
print(set1)
```

**2. clear()** -> Removes all the elements from the set

```python
# Remove all elements from the set

set1 = {"apple", "banana", "cherry"}
set1.clear()
print(set1)
```

**3. difference()** -> Returns a set containing the difference between two or more sets

```python
# Return a set that contains the items that only exist in set 1 and not in set 2

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1.difference(set2)
print(set3)

print("-------------------------")

# Use '-' as a shortcut instead of difference()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1 - set2
print(set3)

print("-------------------------")

# Join more than two sets:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set4 = set1.difference(set2, set3)
print(set4)

print("-------------------------")

# Join more than two sets with the '-' operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set4 = set1 - set2 - set3
print(set4)

print("-------------------------")

# Return a set that contains the items that only exist in set 2, and not in set 1

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set2.difference(set1)
print(set3)
```

**4. difference_update()** -> Removes the items in this set that are also included in another, specified set

```python
# Remove the items that exist in both sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1.difference_update(set2)
print(set1)

print("-------------------------")

# Use '-=' as a shortcut instead of difference_update()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1 -= set2
print(set1)

print("-------------------------")

# Join more than two sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set1.difference_update(set2, set3)
print(set1)

print("-------------------------")

# Join more than two sets with the '-=' operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set1 -= set2 | set3
print(set1)
```

**5. discard()** -> Remove the specified item

```python
# Remove "banana" from the set

set1 = {"apple", "banana", "cherry"}
set1.discard("banana")

print(set1)
```

**6. intersection()** -> Returns a set, that is the intersection of two other sets

```python
# Return a set that contains the items that exist in both set 1 and set set 2

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1.intersection(set2)
print(set3)

print("-------------------------")

# Use '&' as a shortcut instead of intersection()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1 & set2
print(set3)

print("-------------------------")

# Join 3 sets, and return a set with items that is present in all 3 sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cherry", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set4 = set1.intersection(set2, set3)
print(set4)

print("-------------------------")

# Join 3 sets with the '&' operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cherry", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set4 = set1 & set2 & set3
print(set4)
```

**7. intersection_update()** -> Removes the items in this set that are not present in other, specified set(s)

```python
# Remove the items that is not present in both set 1 and set 2

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1.intersection_update(set2)
print(set1)

print("-------------------------")

# Use '&=' as a shortcut instead of intersection_update()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1 &= set2
print(set1)

print("-------------------------")

# Compare 3 sets, and return a set with items that is present in all 3 sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cherry", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set1.intersection_update(set2, set3)
print(set1)

print("-------------------------")

# Join 3 sets with the '&=' operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cherry", "microsoft"}
set3 = {"cherry", "cat", "panda"}

set1 &= set2 & set3
print(set1)
```

**8. isdisjoint()**	-> Returns True if NO items of this set is present in another set

```python
# Return True if no items in set 1 is present in set 2

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "instagram", "microsoft"}

set3 = set1.isdisjoint(set2)
print(set3)

print("-------------------------")

# What if one or more items are present in both sets?

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1.isdisjoint(set2)
print(set3)
```

**9. issubset()**

<= -> Returns True if all items of this set is present in another set

<  -> Returns True if all items of this set is present in another, larger set

```python
# Return True if all items in set 1 are present in set 2

set1 = {"apple", "banana", "cherry"}
set2 = {"apple", "google", "banana", "instagram", "microsoft", "cherry"}

set3 = set1.issubset(set2)
print(set3)

print("-------------------------")

# Use '<=' as a shortcut instead of issubset()

set1 = {"apple", "banana", "cherry"}
set2 = {"apple", "google", "banana", "instagram", "microsoft", "cherry"}

set3 = set1 <= set2
print(set3)

print("-------------------------")

# What if not all items are present in the specified set? Return False if not all items in set 1 are present in set 2

set1 = {"apple", "banana", "cherry"}
set2 = {"apple", "google", "banana", "instagram", "microsoft"}

set3 = set1.issubset(set2)
print(set3)
```

**10. issuperset()**

>= -> Returns True if all items of another set is present in this set

>  -> Returns True if all items of another, smaller set is present in this set

```python
# Return True if all items set 2 are present in set 1

set1 = {"apple", "google", "banana", "instagram", "microsoft", "cherry"}
set2 = {"apple", "banana", "cherry"}

set3 = set1.issuperset(set2)
print(set3)

print("-------------------------")

# Use '>=' as a shortcut instead of issuperset()

set1 = {"apple", "google", "banana", "instagram", "microsoft", "cherry"}
set2 = {"apple", "banana", "cherry"}

set3 = set1 >= set2
print(set3)

print("-------------------------")

# What if not all items are present in the specified set? Return False if not all items in set 2 are present in set 1

set1 = {"apple", "google", "banana", "instagram", "microsoft"}
set2 = {"apple", "banana", "cherry"}

set3 = set1.issuperset(set2)
print(set3)
```

**11. pop()** -> Removes an element from the set

```python
# Remove a random item from the set

set1 = {"apple", "banana", "cherry"}

set1.pop()
print(set1)

print("-------------------------")

# Return the removed element

set1 = {"apple", "banana", "cherry"}

set2 = set1.pop()
print(set2)
```

**12. remove()** -> Removes the specified element

```python
# Remove "banana" from the set

set1 = {"apple", "banana", "cherry"}

set1.remove("banana")
print(set1)
```

**13. symmetric_difference()** -> Returns a set with the symmetric differences of two sets

```python
# Return a set that contains all items from both sets, except items that are present in both sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1.symmetric_difference(set2)
print(set3)

print("-------------------------")

# Use '^' as a shortcut instead of symmetric_difference()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1 ^ set2
print(set3)
```

**14. symmetric_difference_update()** -> Inserts the symmetric differences from this set and another

```python
# Remove the items that are present in both sets, AND insert the items that is not present in both sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1.symmetric_difference_update(set2)
print(set1)

print("-------------------------")

# Use '^=' as a shortcut instead of symmetric_difference_update()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1 ^= set2
print(set1)
```

**15. union()** -> Return a set containing the union of sets

```python
# Return a set that contains all items from both sets, duplicates are excluded

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1.union(set2)
print(set3)

print("-------------------------")

# Use '|' as a shortcut instead of union()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set3 = set1 | set2
print(set3)

print("-------------------------")

# Unify more than 2 sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cat", "microsoft"}
set3 = {"cat", "apple", "panda"}

set4 = set1.union(set2, set3)
print(set4)

print("-------------------------")

# Unify 3 sets with the | operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cat", "microsoft"}
set3 = {"cat", "apple", "panda"}

set4 = set1 | set2 | set3
print(set4)
```

**16. update()** -> Update the set with the union of this set and others

```python
# Insert the items from set 2 into set 1

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1.update(set2)
print(set1)

print("-------------------------")

# Use '|=' as a shortcut instead of update()

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "apple", "microsoft"}

set1 |= set2
print(set1)

print("-------------------------")

# Insert more than one set

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cat", "microsoft"}
set3 = {"cat", "apple", "panda"}

set1.update(set2, set3)
print(set1)

print("-------------------------")

# Join more than one set with the |= operator

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "cat", "microsoft"}
set3 = {"cat", "apple", "panda"}

set1 |= set2 | set3
print(set1)
```
