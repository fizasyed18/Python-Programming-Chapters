### **Loops**

Loops are used to execute a block of code repeatedly until a condition is met or all items in a sequence are processed. The main types are:
- For loops (iterating over sequences)
- While loops (executing code based on a condition).

**1. For Loops**

A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

```python
n = 4

for n in range(0, n):
    print(n)
```

```python
# Print each fruit in a fruit list

fruits = ["apple", "banana", "cherry", "mango"]

for i in fruits:
    print(i)
```

**Looping Through a String**

Even strings are iterable objects, they contain a sequence of characters.

```python
# Loop through the letters in the word "banana"

fruits = ["apple", "banana", "cherry", "mango"]

for i in "banana":
    print(i)
```

**The break Statement**

With the break statement we can stop the loop before it has looped through all the items.

```python
# Exit the loop when x is "banana"

fruits = ["apple", "banana", "cherry", "mango"]

for i in fruits:
    print(i)
    
    if i == "banana":
        break
```

```python
# Exit the loop when x is "banana", but this time the break comes before the print:

fruits = ["apple", "banana", "cherry", "mango"]

for i in fruits:
    if i == "banana":
        break

    print(i)
```

**The continue Statement**

With the continue statement we can stop the current iteration of the loop, and continue with the next.

```python
# Do not print banana

fruits = ["apple", "banana", "cherry", "mango"]

for i in fruits:
    if i == "banana":
        continue
    
    print(i)
```

**The range() Function**

To loop through a set of code a specified number of times, we can use the range() function. The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

```python
# Using the range() function

for i in range(10):
    print(i)
```

```python
# Using the start parameter

for i in range(2, 10):
    print(i)
```

```python
# Increment the sequence with 3 (default is 1)

for i in range(2, 20, 3):
    print(i)
```

**Else in For Loop**

The else keyword in a for loop specifies a block of code to be executed when the loop is finished.

```python
# Print all numbers from 0 to 5, and print a message when the loop has ended

for i in range(6):
    print(i)
    
else:
    print("Finally finished!")
```

```python
# Break the loop when x is 3, and see what happens with the else block

for i in range(10):
    if i == 3:
        break
    print(i)
else:
    print("Finally finished!")
```

**Nested Loops**

A nested loop is a loop inside a loop. The "inner loop" will be executed one time for each iteration of the "outer loop".

```python
# Print each adjective for every fruit

adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for i in adj:
    for j in fruits:
        print(i, j)
```

**The pass Statement**

for loops cannot be empty, but if you for some reason have a for loop with no content, put in the pass statement to avoid getting an error.

```python
for i in [0, 1, 2]:
    pass
```

**While Loop**

While loop repeatedly executes a block of code as long as the given condition remains true. When the condition becomes false, the line immediately after the loop in the program is executed.

```python
# Print i as long as i is less than 6

i = 1

while i < 6:
    print(i)
    i = i + 1
```

**The break Statement**

With the break statement we can stop the loop even if the while condition is true.

```python
# Exit the loop when i is 3

i = 1

while i < 6:
    print(i)
    if i == 3:
        break
    i = i + 1
```

**The continue Statement**

With the continue statement we can stop the current iteration, and continue with the next.

```python
# Continue to the next iteration if i is 3:

i = 0

while i < 7:
    i = i + 1
    if i == 3:
        continue
    print(i)
```

**The else Statement**

With the else statement we can run a block of code once when the condition no longer is true.

```python
# Print a message once the condition is false

i = 1

while i < 6:
    print(i)
    i = i + 1
else:
    print("i is no longer less than 6")
```

**Nested Loops**

A nested loop is a loop inside another loop. The inner loop executes completely for every iteration of the outer loop.

```python
for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()
```
