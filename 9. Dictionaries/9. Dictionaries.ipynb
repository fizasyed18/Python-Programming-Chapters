### **Dictionaries**

Dictionaries are used to store data values in `key:value` pairs. A dictionary is a collection which is ordered*, changeable and do not allow duplicates.

**Creating a Dictionary**

```python
# Create and print a dictionary

dict1 = {"name" : "Alice", "age" : 26}
print(dict1)

print("----------------------")

# Using the dict() Constructor

dict1 = dict(name = "Alice", age = 26)
print(dict1)
```

**Dictionary Length**

```python
dict1 = {"name" : "Alice", "age" : 26}
print(len(dict1))
```

**Accessing Dictionary Items**\
A value in a dictionary is accessed by using its key. This can be done either with square brackets [ ] or with the get() method.

```python
# Get the value of the "model" key

# Using Square Brackets

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1["model"]
print(x)

print("----------------------")

# Using get() method

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.get("model")
print(x)
```

- **Get Keys**\
The keys() method will return a list of all the keys in the dictionary.

```python
# Add a new item to the original dictionary, and see that the keys list gets updated as well

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.keys()
print(x)                    # Before the change

print("----------------------")

dict1["color"] = "Black"
print(x)                    # After the change
```

- **Get Values**\
The values() method will return a list of all the values in the dictionary.

```python
# Make a change in the original dictionary, and see that the values list gets updated as well

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.values()
print(x)                    # Before the change

print("----------------------")

dict1["year"] = 2022
print(x)                    # After the change
```

```python
# Add a new item to the original dictionary, and see that the values list gets updated as well

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.values()
print(x)                    # Before the change

print("----------------------")

dict1["color"] = "red"
print(x)                    # After the change
```

- **Get Items**\
The items() method will return each item in a dictionary, as tuples in a list.

```python
# Make a change in the original dictionary, and see that the items list gets updated as well

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.items()
print(x)                # Before the change

print("----------------------")

dict1["year"] = 2020
print(x)                # After the change
```

```python
# Add a new item to the original dictionary, and see that the items list gets updated as well

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.items()
print(x)

print("----------------------")

dict1["color"] = "red"
print(x)
```

- **Check if Key Exists**

```python
# Check if "model" is present in the dictionary

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

if "model" in dict1:
    print("Yes, model is one of the keys in dict1")
```

**Remove Dictionary Items**\
There are several methods to remove items from a dictionary

- **pop() Method**

```python
dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1.pop("model")
print(dict1)
```

- **popitem() Method**

```python
dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1.popitem()
print(dict1)
```

- **del Method**

```python
# The del keyword removes the item with the specified key name

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

del dict1["model"]
print(dict1)
```

```python
# The del keyword can also delete the dictionary completely

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

del dict1
#print(dict1)           ##this will cause an error because "thisdict" no longer exists.
```

- **clear() Method**

```python
dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1.clear()
print(dict1)
```

**Loop Dictionaries**\
You can loop through a dictionary by using a for loop. When looping through a dictionary, the return value are the keys of the dictionary, but there are methods to return the values as well.

```python
# Print all key names in the dictionary

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

for x in dict1:
    print(x)
    
print("----------------------")

# Print all key names in the dictionary - Using keys() method

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

for x in dict1.keys():
    print(x)
```

```python
# Print all values in the dictionary

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

for x in dict1:
    print(dict1[x])
    
print("----------------------")

# Print all values in the dictionary - Using values() method

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

for x in dict1.values():
    print(x)
```

```python
# Loop through both keys and values, by using the items() method

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

for x, y in dict1.items():
    print(x, y)
```

**Nested Dictionaries**\
A dictionary can contain dictionaries, this is called nested dictionaries.

```python
# Create a dictionary that contain three dictionaries

cars = {
    "car1" : {"brand":"BMW", "color":"white"},
    "car2" : {"brand":"Ford", "color":"red"},
    "car3" : {"brand":"Porsche", "color":"black"}
}

print(cars)
```

```python
# Create three dictionaries, then create one dictionary that will contain the other three dictionaries

car1 = {"brand":"BMW", "color":"white"}
car2 = {"brand":"Ford", "color":"red"}
car3 = {"brand":"Porsche", "color":"black"}

cars = {"car1":car1, "car2":car2, "car3":car3}
print(cars)
```

- **Access Items in Nested Dictionaries**\
To access items from a nested dictionary, you use the name of the dictionaries, starting with the outer dictionary

```python
# Print the brand of car 2

cars = {
    "car1" : {"brand":"BMW", "color":"white"},
    "car2" : {"brand":"Ford", "color":"red"},
    "car3" : {"brand":"Porsche", "color":"black"}
}

print(cars["car2"]["brand"])
```

- **Loop Through Nested Dictionaries**

```python
# Loop through the keys and values of all nested dictionaries

cars = {
    "car1" : {"brand":"BMW", "color":"white"},
    "car2" : {"brand":"Ford", "color":"red"},
    "car3" : {"brand":"Porsche", "color":"black"}
}

for x, obj in cars.items():
    print(x)
    
    for y in obj:
        print(y + ':', obj[y])
```

### **Dictionary Methods**

**1. clear()** -> Removes all the elements from the dictionary

```python
# Remove all elements from the dictionary

dict1 = {"brand" : "Ford", "color" : "red", "year": 1964}

dict1.clear()
print(dict1)
```

**2. fromkeys()** -> Returns a dictionary with the specified keys and value

```python
# Create a dictionary with 3 keys, all with the value 0

x = ('key1', 'key2', 'key3')
y = 0

dict1 = dict.fromkeys(x, y)
print(dict1)
```

**3. get()** -> Returns the value of the specified key

```python
# Get the value of the "model" item

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.get("model")
print(x)

print("----------------------")

# Return the value of an item that do not exist

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.get("price", 3000)
print(x)
```

**4. items()** -> Returns a list containing a tuple for each key value pair

```python
# Return the dictionary's key-value pairs

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.items()
print(x)

print("----------------------")

# When an item in the dictionary changes value, the view object also gets updated

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1["year"] = 2020
print(dict1)
```

**5. keys()** -> Returns a list containing the dictionary's keys

```python
# Return the keys

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.keys()
print(x)

print("----------------------")

# When an item is added in the dictionary, the view object also gets updated

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.keys()
dict1["color"] = "red"
print(x)
```

**6. pop()** -> Removes the element with the specified key

```python
# Remove "model" from the dictionary

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1.pop("model")
print(dict1)

print("----------------------")

# The value of the removed item is the return value of the pop() method

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.pop("model")
print(x)
```

**7. popitem()** -> Removes the last inserted key-value pair

```python
# Remove the last item from the dictionary

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1.popitem()
print(dict1)

print("----------------------")

# The removed item is the return value of the popitem() method

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.popitem()
print(x)
```

**8. update()** -> Updates the dictionary with the specified key-value pairs

```python
# Insert an item to the dictionary

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

dict1.update({"color":"white"})
print(dict1)
```

**9. values()** -> Returns a list of all the values in the dictionary

```python
# Return the values

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.values()
print(x)

print("----------------------")

# When a values is changed in the dictionary, the view object also gets updated

dict1 = {"brand" : "Ford", "model" : "Mustang", "year": 1964}

x = dict1.values()
dict1["year"] = 2018
print(dict1)
```
