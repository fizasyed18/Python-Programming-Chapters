### **Functions**

Python functions are reusable blocks of code used to perform a specific task. They help organize programs into smaller sections and execute the same logic whenever needed by calling the function.

**Creating a Function**

A function is defined using the def keyword, followed by a function name and parentheses.

```python
def myfunction():
    print("Hello World!")
```

**Calling a Function**

```python
def myfunction():
    print("Hello World")
    
myfunction()
```

**Function Names**

Function names follow the same rules as variable names in Python:

1. A function name must start with a letter or underscore
2. A function name can only contain letters, numbers, and underscores
3. Function names are case-sensitive (myFunction and myfunction are different)

Valid function names:

- calculate_sum()
- _private_function()
- myFunction2()

**Why Use Functions?**

Imagine you need to convert temperatures from Fahrenheit to Celsius several times in your program. Without functions, you would have to write the same calculation code repeatedly:

```python
# Without functions - repetitive code:

temp1 = 77
celsius1 = (temp1 - 32) * 5 / 9
print(celsius1)

temp2 = 95
celsius2 = (temp2 - 32) * 5 / 9
print(celsius2)

temp3 = 50
celsius3 = (temp3 - 32) * 5 / 9
print(celsius3)
```

```python
# With functions - reusable code:

def fahrenheit_to_celcius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celcius(77))
print(fahrenheit_to_celcius(95))
print(fahrenheit_to_celcius(50))
```

**Return Values**

Functions can send data back to the code that called them using the return statement. When a function reaches a return statement, it stops executing and sends the result back:

```python
# A function that returns a value
def get_greeting():
    return "Hello World"

message = get_greeting()
print(message)
```

```python
# Using the return value directly

def get_greeting():
    return "Hello World"

print(get_greeting())
```

**The pass Statement**

Function definitions cannot be empty. If you need to create a function placeholder without any code, use the pass statement

```python
def myfunc():
    pass
```

```python
def myfunc(name):           # name is a parameter
    print("Hello", name)
    
myfunc("Hannah")            # "Hannah" is an argument
```

**Parameters vs Arguments**

- A `parameter` is the variable listed inside the parentheses in the function definition.

- An `argument` is the actual value that is sent to the function when it is called.

**Number of Arguments**

By default, a function must be called with the correct number of arguments. If your function expects 2 arguments, you must call it with exactly 2 arguments.

```python
# This function expects 2 arguments, and gets 2 arguments

def myfunc(fname, lname):
    print(fname + " " + lname)
    
myfunc("Hannah", "Baker")
```

```python
# This function expects 2 arguments, but gets only 1

def myfunc(fname, lname):
    print(fname + " " + lname)
    
# myfunc("Hannah")

# If you try to call the function with the wrong number of arguments, you will get an error:
```

**Types of Function Arguments**

Python supports different types of arguments that can be passed during a function call.

**1. Default argument:**\
Default argument use a predefined value when no value is passed during the function call.

**2. Keyword Arguments:**\
Pass values using parameter names, so argument order does not matter.

**3. Positional Arguments:**\
Values are assigned to parameters based on their order in the function call.

**4. Arbitrary Arguments:**\
Allow functions to accept multiple values. This is done using two special symbols:

- *args collects extra positional arguments as a tuple.
- **kwargs collects extra keyword arguments as a dictionary.




**1. Default argument**

```python
def myfunc(x, y = 50):
    print("x:", x)
    print("y:", y)
    
myfunc(10)

# Explanation:

# y=50 sets a default value for parameter y and myFun(10) passes only one argument.
# Since y is not provided, it uses the default value 50.
```

**2. Keyword Arguments**

```python
def student(fname, lname):
    
    print(fname, lname)
    
student(fname= 'Python', lname= 'Fundamentals')
student(lname= 'Fundamentals', fname= 'Python')

# Explanation: 

# fname and lname are passed using parameter names and arguments can be provided in any order.
```

```python
def myfunc(animal, name):
    print("I have a", animal)
    print("My", animal + "'s name is", name)
    
myfunc(animal= "cat", name= "Bella")
```

**3. Positional Arguments**

```python
def student(name, age):
    print("My name is", name)
    print("My age is", age)
    
print("Case 1:")
student("Hannah", 27)

print("Case 2:")
student(27, "Hannah")

# Explanation:

# In Case-1, values match the correct parameters.
# In Case-2, values are swapped because the order changed.
```

```python
def myfunc(animal, name):
    print("I have a", animal)
    print("My", animal + "'s name is", name)
    
myfunc("Cat", "Bella")
```

**4. Arbitrary Arguments**

**Arbitrary Arguments - `*args`**

```python
# Accessing individual arguments from *args:

def myfunc1(*args):
    print("Type:", type(args))
    print("First Argument:", args[0])
    print("Second Argument:", args[1])
    print("All Argument:", args)
    
myfunc1("Hannah", "Alex", "Emily")
```

```python
# Using *args with Regular Arguments

def myfunc(greeting, *names):
    for name in names:
        print(greeting, name)
        
myfunc("Hello", "Hannah", "Alex", "Emily")
```

**Practical Example with `*args`**

```python
# A function that calculates the sum of any number of values:

def myfunc(*numbers):
    total = 0
    for num in numbers:
        total = total + num
    return total

print(myfunc(1, 2, 3))
print(myfunc(10, 20, 30, 40))
```

```python
# Finding the maximum value:

def myfunc(*numbers):
    if len(numbers) == 0:
        return None
    
    max_num = numbers[0]
    
    for num in numbers:
        if num > max_num:
            max_num = num
            
    return max_num
    
print(myfunc(3, 7, 9, 5))
print(myfunc(36, 67, 29, 50))
```

**Arbitrary Keyword Arguments - `**kwargs`**

```python
# Accessing values from **kwargs:

def myfunc(**myvar):
    print("Type:", type(myvar))
    print("Name:", myvar["name"])
    print("Age:", myvar["age"])
    print("All Data:", myvar)
    
myfunc(name= "Hannah", age= 27, city= "Berlin")
```

```python
# Using **kwargs with Regular Arguments

def my_func(username, **details):
  print("Username:", username)
  print("Additional details:")
  for key, value in details.items():
    print(" ", key + ":", value)

my_func("emily123", age = 25, city = "Tokyo", hobby = "Crochet")
```

**Combining `*args` and `**kwargs`**

You can use both *args and **kwargs in the same function. The order must be:

1. regular parameters
2. *args
3. **kwargs

```python
def myfunc(title, *args, **kwargs):
    print("Title:", title)
    print("Positional Arguments:", args)
    print("Keyword Arguments:", kwargs)

myfunc("User Info", "Hannah", "Baker", age= 27, city= "Berlin")

# Explanation:

# *args stores extra positional arguments.
# **kwargs stores extra keyword arguments.
# loop prints all positional and keyword values separately.
```

**Unpacking Arguments**

**Unpacking Lists with `*`**

```python
# Using * to unpack a list into arguments

def myfunc(a, b, c):
    return a + b + c

numbers = [1, 2, 3]
result = myfunc(*numbers)
print(result)
```

**Unpacking Dictionaries with `**`**

```python
# Using ** to unpack a dictionary into keyword arguments

def myfunc(fname, lname):
    print("Hello", fname, lname)
    
person_info = {"fname":"Hannah", "lname":"Baker"}
myfunc(**person_info)
```
