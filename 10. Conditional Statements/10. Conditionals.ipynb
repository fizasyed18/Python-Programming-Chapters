### **Conditional Statements**

Conditional statements are used to control the flow of execution in a program based on specific conditions. They allow programs to execute different blocks of code depending on whether a condition evaluates to True or False.

**If Statement**

`if` statement is used to execute a block of code only when a specified condition evaluates to True.

```python
a = 50
b = 89

if b > a:
    print("b is greater than a")
    
print("---------------------------")

age = 26
if age >= 18:
    print("Eligible to vote")
    
print("---------------------------")

num = 16
if num > 0:
    print("The number is positive.")
```

- **Multiple Statements in If Block**\
You can have multiple statements inside an if block. All statements must be indented at the same level.

```python
age = 26

if age >= 18:
    print("You are an adult.")
    print("You are eligible to vote.")
    print("You have full rights.")
```

- **Using Variables in Conditions**\
Boolean variables can be used directly in if statements without comparison operators.

```python
is_logged_in = True

if is_logged_in:
    print("Welcome Back!")
```

- **Shorthand If**\
If you have only one statement to execute, you can put it on the same line as the if statement.

```python
a = 6
b = 2

if a > b: print("a is greater than b")
```

**Elif Statement**

The `elif` keyword allows you to check multiple expressions for True and execute a block of code as soon as one of the conditions evaluates to True.

`elif` is used when you have multiple mutually exclusive conditions to check. This is more efficient than using multiple separate if statements because Python stops checking once it finds a true condition.

```python
a = 33
b = 33

if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
```

- **Multiple Elif Statements**

```python
# Testing multiple conditions

score = 78

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
elif score >= 60:
    print("Grade D")
```

```python
# Categorizing age groups

age = 78

if age < 13:
    print("Child")
elif age < 20:
    print("Teenager")
elif age < 65:
    print("Adult")
elif age >= 65:
    print("Senior")
```

**Else Statement**

The else statement is executed when the if condition (and any elif conditions) evaluate to False.

- **Else without Elif**

```python
a = 200
b = 56

if b > a:
    print("b is greater than a")
else:
    print("a is greater than b")
```

- **Else with Elif**

```python
a = 200
b = 56

if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b arre equal")
else:
    print("a is greater than b")
```

**If else Statement**

If Else statement is used to execute one block of code when the condition is True and another block when the condition is False.

```python
num = 10

if num > 0:
    print("The number is positive.")
else:
    print("The number is negative.")
    
print("---------------------------")

a = 300
b = 56

if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")

print("---------------------------")

# Checking even or odd numbers

num = 7

if num % 2 == 0:
    print("Number is even")
else:
    print("Number is odd")
```

- **Short Hand If ... Else**\
If you have one statement for if and one for else, you can put them on the same line using a conditional expression.

```python
a = 35
b = 244

print("a is greater") if a > b else print("b is greater")
```

**If-elif-else Statement**

elif statement is used to check multiple conditions in a program. It executes a block of code when its condition evaluates to True after previous conditions evaluate to False.

```python
temp = 32

if temp > 30:
    print("It's hot outside")
elif temp > 20:
    print("It's warm outside")
elif temp > 10:
    print("It's cool outside")
else:
    print("It's cold outside")
```

**Logical Operators**

Logical operators are used to combine conditional statements. Python has three logical operators:

- **and** - Returns True if both statements are true
- **or** - Returns True if one of the statements is true
- **not** - Reverses the result, returns False if the result is true

**1. The `and` Operator**\
The and keyword is a logical operator, and is used to combine conditional statements. Both conditions must be true for the entire expression to be true.

```python
# Test if a is greater than b, AND if c is greater than a

a = 240
b = 33
c = 680

if a > b and c > a:
    print("Both conditions are True")
```

**2. The `or` Operator**\
The or keyword is a logical operator, and is used to combine conditional statements. At least one condition must be true for the entire expression to be true.

```python
# Test if a is greater than b, OR if a is greater than c

a = 240
b = 33
c = 680

if a > b or a > c:
  print("One of the conditions is True")
```

**3. The `not` Operator**\
The not keyword is a logical operator, and is used to reverse the result of the conditional statement.

```python
# Test if a is NOT greater than b

a = 38
b = 250

if not a > b:
  print("a is NOT greater than b")
```

- **Combining Multiple Operators**\
You can combine multiple logical operators in a single expression. Python evaluates not first, then and, then or.

```python
# Combining and, or, and not

age = 25
is_student = False
has_discount_code = True

if (age < 18 or age > 65) and not is_student or has_discount_code:
  print("Discount applies!")
```

**Nested If**

if statements inside if statements are called nested if statements.

```python
x = 76

if x > 10:
  print("Number is above ten")
  if x > 20:
    print("Number is also above 20!")
  else:
    print("Number is not above 20.")
```

```python
age = 27
has_license = True

if age >= 18:
    if has_license:
        print("You can drive")
    else:
        print("You need a license")
else:
    print("You are too young for driving")
```

- **Multiple Levels of Nesting**

```python
score = 65
attendance = 90
submitted = True

if score>= 60:
    if attendance >= 80:
        if submitted:
            print("Pass with good standing")
        else:
            print("Pass but missing assignment")
    else:
        print("pass but low attendance")
else:
    print("Fail")
```

- **Nested If vs Logical Operators**\
Sometimes nested if statements can be simplified using logical operators like and.

```python
temp = 25
is_sunny = True

if temp > 20:
  if is_sunny:
    print("Perfect beach weather!")
  
print("---------------------------")

temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
  print("Perfect beach weather!")
```
