### **Tuples**

A tuple is an immutable ordered collection of elements and are ordered, unchangeable, and allow duplicate values.

- **Ordered**\
When we say that tuples are ordered, it means that the items have a defined order, and that order will not change.

- **Unchangeable**\
Tuples are unchangeable, meaning that we cannot change, add or remove items after the tuple has been created.

- **Allow Duplicates**\
Since tuples are indexed, they can have items with the same value

**Creating a Tuple**

A tuple is created by placing all the items inside parentheses (), separated by commas. A tuple can have any number of items.

```python
tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1)
```

```python
# Creating a Tuple with Mixed Datatypes

tuple1 = (7, 'Hello', 8.9, True, {'name': 'John'}, [1, 4, 6])
print(tuple1)
```

**Accessing of Tuples**

```python
# Print the second item in the tuple

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1[1])

print("----------------------------")

# Print the last item of the tuple

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1[-1])

print("----------------------------")

# Return the third, fourth, and fifth item:

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1[2:5])

print("----------------------------")

# Return the items from the beginning to, but NOT included, "kiwi"

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1[:4])

print("----------------------------")

# Returns the items from "cherry" and to the end

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1[1:])

print("----------------------------")

# Return the items from index -4 (included) to index -1 (excluded)

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
print(tuple1[-4:-1])

print("----------------------------")

# Check if "apple" is present in the tuple

tuple1 = ("apple", "cherry", "banana", "orange", "kiwi", "melon", "mango")
if "apple" in tuple1:
    print("Apple is present in the tuple")
```

**Update Tuples**

- **Change Tuple Values**

Once a tuple is created, you cannot change its values. Tuples are unchangeable, or immutable. You can convert the tuple into a list, change the list, and convert the list back into a tuple.

```python
# Convert the tuple into a list to be able to change it

x = ("apple", "cherry", "banana", "orange")
y = list(x)
print(y)            # Tuple changed into List

y[1] = "kiwi"
x = tuple(y)

print(x)            # List changed into Tuple again
```

- **Add Items**

```python
# Convert the tuple into a list, add "kiwi", and convert it back into a tuple

x = ("apple", "cherry", "banana", "orange")
y = list(x)
print(y)            # Tuple changed into List

y.append("kiwi")
x = tuple(y)
print(x)            # List changed into Tuple again
```

- **Remove Items**

```python
# Convert the tuple into a list, remove "apple", and convert it back into a tuple

x = ("apple", "cherry", "banana", "orange")
y = list(x)
print(y)            # Tuple changed into List

y.remove("apple")
x = tuple(y)
print(x)            # List changed into Tuple again

print("----------------------------")

# To delete the tuple completely

x = ("apple", "cherry", "banana", "orange")
del x
#print(x)            #this will raise an error because the tuple no longer exists
```

**Unpack Tuples**

- **Unpacking a Tuple**

```python
x = ("apple", "banana", "cherry")
(green, yellow, red) = x

print(green)
print(yellow)
print(red)
```

- **Using Asterisk(*)**

If the number of variables is less than the number of values, you can add an * to the variable name and the values will be assigned to the variable as a list

```python
# Assign the rest of the values as a list called "red"

x = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = x

print(green)
print(yellow)
print(red)

print("----------------------------")

# Add a list of values the "tropic" variable

x = ("apple", "mango", "papaya", "pineapple", "cherry")

(green, *tropic, red) = x

print(green)
print(tropic)
print(red)
```

**Loop Tuples**

- **Loop Through a Tuple**

```python
# Iterate through the items and print the values

x = ("apple", "banana", "cherry")

for item in x:
    print(item)
```

- **Loop Through the Index Numbers**

```python
x = ("apple", "banana", "cherry")

for item in range(len(x)):
    print(x[item])
```

- **Using a While Loop**

```python
x = ("apple", "banana", "cherry")
i = 0

while i < len(x):
    print(x[i])
    i = i + 1
```

**Join Tuples**

- **Join Two Tuples**

```python
tuple1 = ("a", "b", "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

- **Multiply Tuples**

```python
# Multiply the fruits tuple by 2

tuple1 = ("apple", "banana", "cherry")
x = tuple1 * 2

print(x)
```

**Tuple Methods**

- **count()**	-> Returns the number of times a specified value occurs in a tuple

```python
# Return the number of times the value 5 appears in the tuple

tuple1 = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = tuple1.count(5)

print(x)
```

- **index()**	-> Searches the tuple for a specified value and returns the position of where it was found

```python
# Search for the first occurrence of the value 8, and return its position

tuple1 = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = tuple1.index(8)

print(x)

print("----------------------------")

# Using the start parameter: Return the first occurrence of the value 8, starting the search at index 5

tuple1 = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = tuple1.index(8, 5)

print(x)
```
