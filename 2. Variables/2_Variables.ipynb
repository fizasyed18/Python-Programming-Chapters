### **Variables**

- Variables are cotainers for storing data values.

### **Rules to Assign a Variable**

1- The variable should contain letters, digits and underscores

2- The first character cannot be a digit.

3- Spaces are not allowed

4- Python reserved keywords cannot be used naming the variable (break, mean, median, test etc)

5- Variable name should be short and descriptive

6- Case sensitivity (Lowercase, uppercase letters... use lowercase letters)

### **Creating Variables**

```python
name = "Alice"          # Creates a string variable
age = 25                # Creates an integer variable
miles = 165.5           # Creates a float variable
is_student = True       # Creates a boolean variable

print(name)
print(age)
print(miles)
print(is_student)
```

### **Casting** 
- To specify the data type of a variable

```python
x = str(3)    # x will be '3'
y = int(3)    # y will be 3
z = float(3)  # z will be 3.0
```

### **Checking Data Types**
- To get the data type of a variable

```python
name = "Alice"          # Creates a string variable
age = 25                # Creates an integer variable
miles = 165.5           # Creates a float variable
is_student = True       # Creates a boolean variable

print(type(name))
print(type(age))
print(type(miles))
print(type(is_student))
```

### **Variable Name Techniques**

```python
# CAMEL CASE
myVariableName = "Alice"

# PASCAL CASE
MyVariableName = "Alice"

# SNAKE CASE
my_variable_name = "Alice"
```

### **To Assign Multiple Values**

- Many Values to Multiple Variables

```python
x, y, z = "Apple", "Banana", "Cherry"

print(x)
print(y)
print(z)
```

- One Value to Multiple Variables

```python
x = y = z = "Alice"

print(x)
print(y)
print(z)
```

- To Unpack a Collection

```python
names = ["Alice", "Bob", "Charlie"]
x, y, z = names

print(x)
print(y)
print(z)
```

### **Variables with User Input**

```python
name = input("Enter your name:")
print("Hello, " + name + "!")
```

### **Global Variables**

- Variables that are created outside of a function.

- They can be accessed anywhere in the program.

- Changes affect the entire program.

Variable outside of a function, and then use it inside the function

```python
x = "awesome"

def myfunc():
    print("Python is " + x)
    
myfunc()
```

Variable inside a function, with the same name as the global variable

```python
x = "awesome"

def myfunc():
    x = "fantastic"
    print("Python is " + x)
    
myfunc()

print("Python is " + x)
```

### **Local Variables**

- Any variable created outside a function can be accessed within any function

- They are accessible only inside the function.

- Changes are local and don’t affect other functions.

```python
x = 5
y = 10

def sum():
    sum = x + y
    return sum

print(sum())
```
