### **Strings**

Strings are sequence of characters written inside quotes. It can include letters, numbers, symbols and spaces.

- A single character is treated as a string of length one.
- Strings are commonly used for text handling and manipulation.

#### **Syntax of String Slicing**

*substring = s[start:end:step]*

**Parameters:**

s = Original string\
start = Starting index (inclusive)\
end = Stopping index (exclusive)\
step = Interval between indices. A positive value slices from left to right, while a negative value slices from right to left.

### **Creating a String**

Strings can be created using either single ('...') or double ("...") quotes. Both behave the same.

```python
a = 'Python'
b = "Python Fundamentals"

print(a)
print(b)
```

##### **Multi-line Strings**

Use triple quotes ('''...''' ) or ( """...""") for strings that span multiple lines.

```python
a = """I am learning Python."""
b = '''Python is a programming language.'''

print(a)
print(b)
```

### **Accessing Characters in String**

Strings are indexed sequences. Positive indices start at 0 from the left, negative indices start at -1 from the right.

```python
a = "Python"

print(a[0])     # Output: P
print(a[1])     # Output: y
print(a[-1])    # Output: n
print(a[-3])    # Output: h
```

### **String Length**

To get the length of a string, `len()` function is used.

```python
a = "Python"

print(len(a)) 
```

### **String Slicing**

Slicing is a way to extract a portion of a string by specifying the start and end indexes. The syntax for slicing is string[start:end], where start starting index and end is stopping index (excluded).

```python
a = "Python Fundamentals"

print(a[1:3])   
```

##### **1. Slice From the Start**

If we leave out the starting index, the range will start at the first character.

```python
a = "Python Fundamentals"

print(a[:7])  
```

##### **2. Slice To the End**

If we leave out the ending index, the range will go to the end.

```python
a = "Python Fundamentals"

print(a[7:])     
```

##### **3. Negative Indexing**

Negative indexes are used to start the slice from the end of the string.

```python
a = "Python Fundamentals"

print(a[-4:-1])
print(a[-7:-2:2])
print(a[-2:])
print(a[:-3])
```

##### **4. Reverse a String**

```python
a = "Python"

print(a[::-1])
```

### **Looping Through Strings**

Strings are iterable, which means we can access each character one by one using a loop.

```python
for x in "python":
    print(x)
```

```python
a = "Python"

for char in a:
    print(char)
```

### **String Concatenation**

To concatenate, or combine two strings, the `+` operator is used.

```python
a = "Python"
b = "Fundamentals"

c = a + " " + b
print(c)
```

### **Formatting Strings**

##### **1. Using f-strings:** 

f-strings allows to directly insert variables and expressions inside a string using {} brackets.

```python
name = "John"
age = 30

print(f"Name: {name}, Age: {age}")
```

##### **2. Using format:** 

format() method allows inserting values into placeholders {} inside a string.

```python
a = "My name is {} and I'm {} years old.".format("John", 30)

print(a)
```

#####  **Placeholder** 

A placeholder can contain variables, operations, functions, and modifiers to format the value.

```python
price = 800
txt = f"The price is {price} dollars."

print(txt)
```

##### **Modifier** 

A modifier is included by adding a colon : followed by a legal formatting type, like .2f which means fixed point number with 2 decimals.

```python
price = 800
txt = f"The price is {price:.2f} dollars."

print(txt)
```

A placeholder can contain Python code, like math operations:

```python
txt = f"The price is {20 * 47} dollars."

print(txt)
```

### **Updating a String**

Strings cannot be changed directly after creation. Any modification results in a new string being created using slicing or methods like replace().

```python
a = "abcd efgh"

a1 = a.replace("abcd", "ABCD")

print(a1)
```

### **Deleting a String**

Individual characters of a string cannot be deleted because strings are immutable. However, an entire string variable can be removed using `del` keyword.

```python
a = "Python"

del a
```

### **String Methods**

##### **1. capitalize -> Converts the first character to upper case**

```python
txt1 = "python"
print(txt1.capitalize())

print("-------------")

txt2 = "python is a programming language."
print(txt2.capitalize())
```

##### **2. count -> Returns the number of times a specified value occurs in a string**

```python
# Return the number of times the value "Python" appears in the string.

txt1 = "I love Python. Python is great."
print(txt1.count("Python"))

print("-------------")

# Search from position 5 to 15.

txt2 = "I love Python. Python is great."
print(txt2.count("Python", 5, 15))
```

##### **3. startswith -> Returns true if the string starts with the specified value**

```python
# Check if the string starts with "Python"

txt1 = "Python is a programming language."
print(txt1.startswith("Python"))

print("-------------")

# Check if position 7 to 20 starts with the characters "pro"

txt2 = "Python is a programming language."
print(txt2.startswith("pro", 11, 20))

print("-------------")

# Check if the string starts with either "Python" or "language"

txt3 = "Python is a programming language."
print(txt3.startswith(("Python", "Language")))
```

##### **4. endswith -> Returns true if the string ends with the specified value**

```python
# Check if the string ends with punctuation sign (.)"

txt1 = "Python is a programming language."
print(txt1.endswith("."))

print("-------------")

# Check if the string ends with the phrase "programming."

txt2 = "Python is a programming language."
print(txt2.endswith("programming."))

print("-------------")

# Check if the string ends with either the phrase "great" or "language"

txt3 = "Python is a programming language."
print(txt3.endswith(("great", "language.")))

print("-------------")

# Check if position 5 to 11 ends with the phrase "my world."

txt4 = "Python is a programming language."
print(txt4.endswith("language.", 5, 11))
```

##### **5. isupper -> Returns True if all characters in the string are upper case**

```python
# Check if all the characters in the texts are in upper case.

txt1 = "PYTHON IS A PROGRAMMING LANGUAGE."
print(txt1.isupper())

print("-------------")

txt2 = "Python is a programming language."
print(txt2.isupper())

print("-------------")

txt3 = "python is a programming language."
print(txt3.isupper())
```

##### **6. islower -> Returns True if all characters in the string are lower case**

```python
# Check if all the characters in the text are in lower case.

txt1 = "Python is a programming language."
print(txt1.islower())

print("-------------")

txt2 = "python is a programming language."
print(txt2.islower())
```

##### **7. isdigit -> Returns True if all characters in the string are digits**

```python
# Check if all the characters in the text are digits

txt1 = "97870"
print(txt1.isdigit())

print("-------------")

txt2 = "\u0030"             #unicode for 0
print(txt2.isdigit())

print("-------------")

txt3 = "\u00B2"             #unicode for ²
print(txt3.isdigit())
```

##### **8. isnumeric -> Returns True if all characters in the string are numeric**

```python
# Check if all the characters in the text are numeric.

txt1 = "87932"
print(txt1.isnumeric())

print("-------------")

txt2 = "-13"
print(txt2.isnumeric())

print("-------------")

txt3 = "3.14"
print(txt3.isnumeric())

print("-------------")

txt4 = "10km2"
print(txt4.isnumeric())

print("-------------")

txt5 = "\u0030"             #unicode for 0
print(txt5.isdigit())

print("-------------")

txt6 = "\u00B2"             #unicode for ²
print(txt6.isdigit())
```

##### **9. isalpha -> 	Returns True if all characters in the string are in the alphabet**

```python
# Check if all the characters in the text are letters.

txt1 = "Python"
print(txt1.isalpha())

print("-------------")

txt2 = "Python10"
print(txt2.isalpha())
```

##### **10. isascii -> Returns True if all characters in the string are ascii characters**

```python
# Check if all the characters in the text are ascii characters

txt = "Python123"
print(txt.isascii())
```

##### **11. isdecimal -> Returns True if all characters in the string are decimals**

```python
# Check if all the characters in a string are decimals (0-9).

txt1 = "1234567"
print(txt1.isdecimal())

print("-------------")

txt2 = "\u0030"            #unicode for 0
print(txt2.isdecimal())

print("-------------")

txt3 = "\u0047"            #unicode for G   
print(txt3.isdecimal())
```

##### **12. lower -> Converts a string into lower case**

```python
# Lower case the string

txt = "PYTHON"
print(txt.lower())
```

##### **13. upper -> Converts a string into upper case**

```python
# Upper case the string

txt = "python"
print(txt.upper())
```

##### **14. split -> Splits the string at the specified separator, and returns a list**

```python
# Split a string into a list where each word is a list item

txt1 = "Python is a programming language."
print(txt1.split())

print("-------------")

# Split the string, using comma, followed by a space, as a separator

txt2 = "Hello, my name is Peter, I am 26 years old."
print(txt2.split(", "))

print("-------------")

# Use a hash character as a separator

txt3 = "apple#banana#cherry#orange"
print(txt3.split("#"))

print("-------------")

# Split the string into a list with max 2 items

txt4 = "apple#banana#cherry#orange"
print(txt4.split("#", 1))
```

##### **15. strip -> Returns a trimmed version of the string**

```python
# Remove spaces at the beginning and at the end of the string

txt1 = "   Python is a programming language.   "
print(txt1.strip())

print("-------------")

# Remove the leading and trailing characters

txt2 = ",,,,,rrttgg.....banana....rrr"
print(txt2.strip(",.grt"))
```

##### **16. join -> Joins the elements of an iterable to the end of the string**

```python
# Join all items of a tuple into a string, using a hash character as separator

tuple1 = ("apple", "banana", "cherry")
print("#".join(tuple1))

print("-------------")

# Join all items in a dictionary into a string, using the word "TEST" as separator.

dict1 = {"name": "John", "age": 30, "city": "New York"}
print("TEST".join(dict1))
```

##### **17. replace -> Returns a string where a specified value is replaced with a specified value**

```python
# Replace the word "apples" with the word "oranges"

txt1 = "I like apples."
print(txt1.replace("apples", "oranges"))

print("-------------")

# Replace all occurrence of the word "one"

txt2 = "one one was a race horse, two two was one too."
print(txt2.replace("one", "four"))

print("-------------")

# Replace the two first occurrence of the word "one"

txt3 = "one one was a race horse, two two was one too."
print(txt3. replace("one", "four", 2))
```

##### **18. rsplit -> Splits the string at the specified separator, and returns a list**

```python
# Remove any white spaces at the end of the string

txt1 = "    Python    "
print(txt1.rsplit())

print("-------------")

# Remove the trailing characters if they are commas, periods, s, q, or w

txt2 = "banana,,,,,ssqqqww....."
print(txt2.rstrip(",.sqw"))
```

##### **19. rstrip -> Returns a right trim version of the string**

```python
# Split a string into a list, using comma, followed by a space (, ) as the separator.

txt1 = "apple, banana, cherry"
print(txt1.split(", "))

print("-------------")

# Split the string into a list with maximum 2 items

txt2 = "apple, banana, cherry"
print(txt2.split(", ", 1))
```

##### **20. swapcase -> Swaps cases, lower case becomes upper case and vice versa**

```python
txt1 = "Python is a programming language."
print(txt1.swapcase())
```
