### **List**

Lists are used to store multiple items in a single variable and are ordered, changeable, and allow duplicate values. They are dynamic, resizable and capable of storing multiple data types.

- **Mutable:** list elements can be changed, updated, added, or removed after the list is created.
- **Ordered:** elements maintain the order in which they are inserted.
- **Index-based:** elements are accessed using their position, starting from index 0.

**Creating a List**

Lists can be created in several ways, such as using square brackets [], the list() constructor or by repeating elements.

**1. Using Square Brackets**

```python
a = [1, 2, 3, 4]
print(a)

print("--------------")

b = ["apple", "orange"]
print(b)
```

**2. Using list() Constructor**

```python
a = list((1, 2, 3, 'apple', 5.5))
print(a)

print("--------------")

b = list("Python")
print(b)
```

**3. Creating List with Repeated Elements**

```python
a = [5] * 4
print(a)
```

**Accessing Items**

Elements in a list are accessed using indexing. Python uses zero-based indexing, meaning a[0] represents the first element. Negative indexing is also supported, where -1 accesses the last element.

```python
# Print the second item of the list

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]
print(list1[1])
```

```python
# Print the last item of the list

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]
print(list1[-1])
```

```python
# Return the third, fourth, and fifth item

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]
print(list1[2:5])
```

```python
# Return the items from the beginning to, but NOT including, "kiwi"

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]
print(list1[:4])
```

```python
# Return the items from "mango" to the end

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]
print(list1[1:])
```

```python
# Return the items from "mango" (-4) to, but NOT including "kiwi" (-1)

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]
print(list1[-4:-1])
```

```python
# Check if "apple" is present in the list

list1 = ["apple", "mango", "orange", "cherry", "kiwi"]

if "apple" in list1:
    print("Apple is present in the list.")
```

**Change List Items**

```python
# Change the second item

list1 = ["apple", "cherry", "banana", "mango"]
list1[1] = "Orange"

print(list1)
```

```python
# Change the values "cherry" and "banana" with the values "blackcurrant" and "watermelon"

list1 = ["apple", "cherry", "banana", "mango"]
list1[1:3] = ["blackcurrant", "watermelon"]

print(list1)
```

```python
# Change the second value by replacing it with two new values

list1 = ["apple", "cherry", "banana", "mango"]
list1[1:2] = ["blackcurrant", "watermelon"]

print(list1)
```

```python
# Change the second and third value by replacing it with one value

list1 = ["apple", "cherry", "banana", "mango"]
list1[1:3] = ["watermelon"]

print(list1)
```

```python
# Insert "watermelon" as the third item

list1 = ["apple", "cherry", "banana", "mango"]
list1.insert(2, "watermelon")

print(list1)
```

**Add List Items**

Elements can be added to a list using the following methods:

**1. append()** -> Adds an element at the end of the list.

```python
list1 = ["apple", "mango", "orange"]
list1.append("banana")

print(list1)
```

**2. insert()** -> Adds an element at a specific position.

```python
list1 = ["apple", "mango", "orange"]
list1.insert(2, "watermelon")

print(list1)
```

**3. extend()** -> Adds multiple elements to the end of the list.

The extend() method does not have to append lists, you can add any iterable object (tuples, sets, dictionaries etc.).

```python
list1 = ["apple", "mango", "orange"]
list2 = ["pineapple", "papaya"]
list1.extend(list2)

print(list1)

print("-------------------------")

# Add elements of a tuple to a list

list1 = ["apple", "mango", "orange"]
list2 = ("pineapple", "papaya")
list1.extend(list2)

print(list1)
```

**Remove List Items**

Elements can be removed from a list using the following methods:

**1. remove()** -> Removes the first occurrence of an element.

```python
list1 = ["apple", "mango", "orange"]
list1.remove("mango")

print(list1)
```

**2. pop()** -> Removes the element at a specific index or the last element if no index is specified.

```python
# Remove the second item

list1 = ["apple", "mango", "orange"]
list1.pop(1)

print(list1)

print("-------------------------")

# Remove the last item

list1 = ["apple", "mango", "orange"]
list1.pop()

print(list1)
```

**3. del** -> Deletes an element at a specified index.

```python
# Remove the first item

list1 = ["apple", "mango", "orange"]
del list1[0]

print(list1)

print("-------------------------")

# Delete the entire list

list1 = ["apple", "mango", "orange"]
del list1
```

**4. clear()** -> removes all items.

```python
list1 = ["apple", "mango", "orange"]
list1.clear()

print(list1)
```

**Iterating Over Lists**

```python
list1 = ["apple", "mango", "banana"]

for x in list1:
    print(x)
```

- **Using range() with for Loop**

```python
# Print all items by referring to their index number

list1 = ["apple", "mango", "banana"]

for i in range(len(list1)):
    print(list1[i])
```

- **Using enumerate()**

```python
list1 = ["apple", "mango", "banana"]

for i, x in enumerate(list1):
    print(i, x)
```

- **Using a While Loop**

```python
# Print all items, using a while loop to go through all the index numbers

list1 = ["apple", "mango", "banana"]
i = 0

while i < len (list1):
    print(list1[i])
    i = i + 1
```

- **Looping Using List Comprehension**

```python
list1 = ["apple", "mango", "banana"]
[print(i) for i in list1]
```

- **List Comprehension**

```python
# Without list comprehension, you will have to write a for statement with a conditional test inside:

list1 = ["apple", "mango", "banana", "orange", "kiwi"]
newlist = []

for i in list1:
    if "a" in i:
        newlist.append(i)
        
print(newlist)

print("-------------------------")

# With list comprehension, you can do all that with only one line of code:

list1 = ["apple", "mango", "banana", "orange", "kiwi"]
newlist = [i for i in list1 if "a" in i]

print(newlist)
```

```python
# Only accept items that are not "apple"

list1 = ["apple", "mango", "banana", "orange", "kiwi"]
newlist = [i for i in list1 if i != "apple"]

print(newlist)
```

```python
# Accept only numbers lower than 7

newlist = [x for x in range(10) if x < 7]

print(newlist)
```

```python
# Set the values in the new list to upper case

list1 = ["apple", "mango", "banana", "orange", "kiwi"]
newlist = [x.upper() for x in list1]

print(newlist)
```

```python
# Return "cherry" instead of "banana"

list1 = ["apple", "mango", "banana", "orange", "kiwi"]
newlist = [x if x != "banana" else "cherry" for x in list1]

print(newlist)
```

**Sort Lists**

- **Sort List Alphanumerically**

```python
list1 = ["apple", "mango", "banana", "orange", "kiwi"]
list1.sort()
print(list1)

print("-------------------------")

list2 = [300, 34, 85, 96, 290]
list2.sort()
print(list2)
```

- **Sort Descending**

```python
list1 = ["apple", "mango", "banana", "orange", "kiwi"]
list1.sort(reverse= True)

print(list1)

print("-------------------------")

list2 = [300, 34, 85, 96, 290]
list2.sort(reverse= True)

print(list2)
```

- **Case Insensitive Sort**

```python
list1 = ["apple", "mango", "banana", "orange", "kiwi"]
list1.sort(key= str.lower)

print(list1)
```

- **Reverse Order**

```python
list1 = ["apple", "mango", "banana", "orange", "kiwi"]
list1.reverse()

print(list1)
```

**Join Lists**

```python
# Join two list

list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

list3 = list1 + list2
print(list3)
```

```python
# Append list2 into list1

list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

for x in list2:
    list1.append(x)
    
print(list1)
```

```python
# Use the extend() method to add list2 at the end of list1

list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

list1.extend(list2)
print(list1)
```

### **List Methods**

**1. append()** -> Adds an element at the end of the list

```python
# Add an element to the list1

list1 = ["apple", "banana", "cherry"]
list1.append("orange")

print(list1)

print("-------------------------")

# Add a list to a list

list1 = ["apple", "banana", "cherry"]
list2 = ["Ford", "BMW", "Volvo"]

list1.append(list2)
print(list1)
```

**2. count()** -> Returns the number of elements with the specified value

```python
# Return the number of times the value "cherry" appears in the list1

list1 = ["apple", "banana", "cherry"]
x = list1.count("apple")

print(x)

print("-------------------------")

# Return the number of times the value 9 appears int the list

list1 = [1, 4, 2, 9, 7, 8, 9, 3, 1]
x = list1.count(9)

print(x)
```

**3. extend()** -> Add the elements of a list (or any iterable), to the end of the current list

```python
# Add the elements of list1 to the list2

list1 = ["apple", "banana", "cherry"]
list2 = ["Ford", "BMW", "Volvo"]

list1.extend(list2)
print(list1)

print("-------------------------")

# Add a tuple to the fruits list

list1 = ["apple", "banana", "cherry"]
list2 = (1, 4, 5, 8)

list1.extend(list2)
print(list1)
```

**4. index()** -> Returns the index of the first element with the specified value

```python
# What is the position of the value "cherry"

list1 = ["apple", "banana", "cherry"]
x = list1.index("cherry")
print(x)

print("-------------------------")

# What is the position of the value 32

list1 = [4, 55, 64, 32, 16, 32]
x = list1.index(32)
print(x)

print("-------------------------")

# Find the postion of 'cherry', but start the search at position 4

list1 = ["apple", "banana", "cherry", "kiwi", "watermelon", "cherry"]
x = list1.index("cherry", 4)
print(x)
```

**5. insert()** -> Adds an element at the specified position

```python
# Insert the value "orange" as the second element of the list1

list1 = ["apple", "banana", "cherry"]
list1.insert(1, "orange")

print(list1)
```

**6. reverse()**	-> Reverses the order of the list

```python
# Reverse the order of the list1

list1 = ["apple", "banana", "cherry"]
list1.reverse()
print(list1)
```

**7. sort()** -> Sorts the list

```python
# Sort the list alphabetically

list1 = ["apple", "cherry", "mango", "banana"]
list1.sort()

print(list1)

print("-------------------------")

# Sort the list descending

list1 = ["apple", "cherry", "mango", "banana"]
list1.sort(reverse= True)

print(list1)
```

**8. clear()** -> Removes all the elements from the list

```python
# Remove all elements from the list1

list1 = ["apple", "banana", "cherry"]
list1.clear()
```

**9. copy()** -> Returns a copy of the list

```python
# Copy the list1

list1 = ["apple", "banana", "cherry"]
x = list1.copy()
print(x)
```

**10. remove()** -> Removes the item with the specified value

```python
# Remove the "banana" element of the list1

list1 = ["apple", "banana", "cherry"]
list1.remove("banana")

print(list1)
```

**11.  pop()** -> Removes the element at the specified position

```python
# Remove the second element of the list1

list1 = ["apple", "banana", "cherry"]
list1.pop(1)

print(list1)
```
